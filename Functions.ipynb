  1.  What is the difference between a function and a method in Python?

  Answer.

  In Python, a **function** and a **method** are both reusable blocks of code, but the main difference is that a function works independently, while a method is associated with an object or class. A function can be called directly by using its name, for example, `add(10, 20)`. On the other hand, a method is usually called using an object, for example, `numbers.append(4)`, where `append()` is a method of the list object. In simple words, a **function is independent and can be called directly, whereas a method is associated with an object or class and is generally called through that object**.

2. Explain the concept of function arguments and parameters in Python.

Answer.

In Python, **parameters and arguments** are used to pass data to a function. A **parameter** is a variable defined inside the function definition that receives a value. An **argument** is the actual value that we pass to the function when calling it. For example, in `def add(a, b):`, `a` and `b` are parameters, and when we call `add(10, 20)`, `10` and `20` are arguments. Python supports different types of arguments, such as positional arguments, keyword arguments, default arguments, and variable-length arguments. In simple words, **parameters are the variables defined in the function, while arguments are the actual values passed to those parameters when the function is called**.

3.  What are the different ways to define and call a function in Python?

Answer.

In Python, a function is defined using the `def` keyword followed by the function name and parentheses. A function can be defined in different ways depending on whether it requires parameters, returns a value, or uses default or variable-length arguments. For example, a simple function can be defined as `def greet(): print("Hello")` and called using `greet()`. A function with parameters can be defined as `def add(a, b): return a + b` and called using `add(10, 20)`. We can also use default parameters, such as `def greet(name="User")`, and call it with or without an argument. Functions can also accept variable numbers of arguments using `*args` and `**kwargs`. In Python, functions can be called using positional arguments, keyword arguments, or a combination of both. In simple words, **we define functions using `def` and call them by writing the function name followed by parentheses with the required arguments**.

4.  What is the purpose of the `return` statement in a Python function?

Answer.

The `return` statement in Python is used to **send a value or result back from a function to the place where the function was called**. It also stops the execution of the function at that point. For example, in `def add(a, b): return a + b`, the `return` statement sends the sum of `a` and `b` back to the caller. We can store the returned value in a variable, print it, or use it in another expression. If a function does not have a `return` statement, it automatically returns `None`. In simple words, **the `return` statement is used to return the result of a function and end its execution**.

5. What are iterators in Python and how do they differ from iterables?

Answer.

In Python, an **iterable** is an object that can be looped over one element at a time, such as a list, tuple, string, or dictionary. An **iterator** is an object that keeps track of its current position and returns the next element using the `next()` function. We can create an iterator from an iterable using the `iter()` function. For example, `numbers = [1, 2, 3]` is an iterable, while `iterator = iter(numbers)` creates an iterator. We can then use `next(iterator)` to get each element one by one. The main difference is that **an iterable can be converted into an iterator, while an iterator produces elements one at a time and remembers its current position**. An iterator also implements both `__iter__()` and `__next__()` methods.

6. Explain the concept of generators in Python and how they are defined.

Answer.

In Python, a **generator** is a special type of function that generates values **one at a time** instead of creating and storing all the values in memory at once. Generators are useful when working with large amounts of data because they are **memory efficient**. A generator is defined like a normal function using the `def` keyword, but instead of using `return` to produce values, it uses the **`yield`** keyword. When `yield` is used, the function pauses its execution and remembers its current state. When the next value is requested, it continues from where it stopped. For example, `def numbers(): yield 1; yield 2; yield 3` defines a generator function, and it can be called using `numbers()`. In simple words, **a generator produces values one by one using the `yield` keyword, making it useful for handling large or continuous data efficiently**.

7. What are the advantages of using generators over regular functions?

Answer.

Generators have several advantages over regular functions, especially when working with large amounts of data. The main advantage is **memory efficiency**, because a generator produces values one at a time using the `yield` keyword instead of storing all values in memory at once. Generators are also **faster and more efficient** for processing large datasets because they calculate values only when they are needed. They support **lazy evaluation**, which means the next value is generated only when requested. Generators are also useful for handling **large files, streams, and infinite sequences** without consuming a large amount of memory. In simple words, **generators save memory, improve efficiency, and allow data to be processed one value at a time**.

8. What is a lambda function in Python and when is it typically used?

Answer.

A **lambda function** in Python is a small, anonymous function that is defined without using the `def` keyword. It is usually used when we need a simple function for a short period of time. A lambda function can take any number of arguments but can contain only one expression. For example, `square = lambda x: x * x` creates a lambda function that returns the square of a number. Lambda functions are commonly used with functions such as `map()`, `filter()`, and `sorted()` when a simple operation is required. In simple words, **a lambda function is a short, one-line anonymous function used for simple operations without defining a regular function**.

9. Explain the purpose and usage of the `map()` function in Python.

Answer.

The **`map()` function** in Python is used to apply a function to **each element of an iterable**, such as a list, tuple, or set, and return the results as a map object. It is useful when we want to perform the same operation on every element without writing an explicit loop. For example, `numbers = [1, 2, 3, 4]` and `result = map(lambda x: x * 2, numbers)` will multiply each number by 2. We can convert the result into a list using `list(result)`, which gives `[2, 4, 6, 8]`. The `map()` function is commonly used with **lambda functions** for simple transformations. In simple words, **`map()` applies a given function to every element of an iterable and produces the transformed values**.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Answer.

In Python, `map()`, `reduce()`, and `filter()` are commonly used for processing data from an iterable. The **`map()` function** applies a function to every element and returns the transformed values. For example, `map(lambda x: x * 2, numbers)` doubles each number. The **`filter()` function** selects only those elements that satisfy a given condition. For example, `filter(lambda x: x > 5, numbers)` returns only numbers greater than 5. The **`reduce()` function** combines all the elements of an iterable step by step to produce a single final value, such as calculating the sum or product of all elements. `reduce()` is available in the `functools` module. In simple words, **`map()` transforms each element, `filter()` selects elements based on a condition, and `reduce()` combines all elements into one result**.

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

Answer.

class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration

        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1

        return value


n = int(input("Enter number of terms: "))

fib = Fibonacci(n)

for num in fib:
    print(num)

6.  Write a generator function in Python that yields the powers of 2 up to a given exponent.

Answer.

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i


n = int(input("Enter the exponent: "))

    print(value)

7.   Implement a generator function that reads a file line by line and yields each line as a string.

Answer.

def read_file_lines(filename):
    with open(filename, "r") as file:
        for line in file:
            yield line.strip()


# Example usage
for line in read_file_lines("data.txt"):
    print(line)
